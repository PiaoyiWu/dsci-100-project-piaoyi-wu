#### Task 3.1: Tidy the data (5 points)

```r
tidy_womens_mm <- womens_march_madness_data |>
    mutate(top_16_or_higher = tourney_finish %in% c("top_16_loss", "top_8_loss", "top_4_loss", "top_2_loss", "champ")) |> 
    select(conf_wins_pct, reg_wins_pct, top_16_or_higher) |> # first select the rows we need to classify
    # no need to do pivot_longer to tidy as reg pct and conf pct are diff vars that belong to the same team in the same year
    filter(conf_wins_pct != "NA" & reg_wins_pct != "NA") # get rid of NAs

head(tidy_womens_mm)    # this outputs the first 6 rows
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
womens_mm_summary_mean <- womens_march_madness_data |> # pipe from the original data
    select(-c(school, conference, division, bid, first_game_at_home, tourney_finish)) |> # take away the columns that are non-numerical
    map_dfr(mean, na.rm = TRUE) |> # operate/take the mean of all columns but also take away na values
    mutate(summary_type = "mean") # this adds a column to describe what the values on each row mean

womens_mm_summary_min <- womens_march_madness_data |> # use the same as above but instead of the mean you use the min function
    select(-c(school, conference, division, bid, first_game_at_home, tourney_finish)) |>
    map_dfr(min, na.rm = TRUE) |> # using min here instead of mean
    mutate(summary_type = "min") # label the summary type for the observation

womens_mm_summary_max <- womens_march_madness_data |> # same approach, using max instead of mean
    select(-c(school, conference, division, bid, first_game_at_home,  tourney_finish)) |>
    map_dfr(max, na.rm = TRUE) |> # map dr with max rather than mean
    mutate(summary_type = "max") # label


womens_mm_summary <- bind_rows(womens_mm_summary_mean, womens_mm_summary_min, womens_mm_summary_max)
# there is a similar function in the reference sheet which is bind_cols, but this combines rows. (it is also used in the dsci book)
# combines the three dfs together into one with all three rows

womens_mm_summary # output the table
```

**Conference vs. Regional Win Rate**

This plot shows the moderate positive relationship between regional win percentage and conference win percentage. This indicates that teams that do well in conference games generally do well in regional games following. Adding color for the categorical variable could show us how it extrapolates to how teams finish the season.

```r
# option settings carry over

reg_conf_vs_reg_plot <- tidy_womens_mm |>
    ggplot(aes(x = conf_wins_pct, y = reg_wins_pct)) + # they do not predict each other but because conference games happen before
                                                       # regional games, i put conf wins pct as the x axis
    geom_point(alpha = 0.4) +                          # this for readability
    labs(x = "Conference Wins %", y = "Regional Wins %") + # below is labelling
    ggtitle("Conference vs Regional Win Rate") +
    theme(text = element_text(size = 18))

reg_conf_vs_reg_plot
```

**Re. Wins Top 16 or Higher**

The following double histogram shows the distribution of regional win rates and whether they finished top 16 or higher. This is useful for answering our research question as we can clearly see emprically that the center of top 16 finishing teams generally do better (greater win rate) than those that dont. 

```r
# double histogram

reg_wins_pct_top_16_or_not <- tidy_womens_mm |>
    ggplot(aes(x = reg_wins_pct, fill = top_16_or_higher)) + # fill changes the histogram to separate values by color
    geom_histogram(bins = 10, position = "identity", alpha = 0.6) + # prevents value from sitting on top and also makes it translucent
    scale_fill_brewer(palette = "Set1") +                          # red and blue for more contrast
    labs(x = "Regional Wins %", y = "Num. Teams", fill = "Top 16 or Higher") + # axis and labels
    ggtitle("Reg. Wins Top 16 or Higher") +
    theme(text = element_text(size = 20))

reg_wins_pct_top_16_or_not
```
